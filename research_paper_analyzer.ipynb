```python
import PyPDF2
import os
import re
```

```python
def extract_text_from_pdf(pdf_path):
    """
    Extracts all text from a PDF file page by page.
    Returns the full text as a single string.
    """
    text = ""
    try:
        with open(pdf_path, "rb") as file:
            reader = PyPDF2.PdfReader(file)
            total_pages = len(reader.pages)
            print(f"  Total pages: {total_pages}")

            for page in reader.pages:
                page_text = page.extract_text()
                if page_text:
                    text += page_text + "\n"

        print(f"  ✅ Extracted {len(text)} characters")

    except Exception as e:
        print(f"  ❌ Error: {e}")

    return text
```

 Text Cleaning Function

```python
import re

def clean_text(text):
    """
    Cleans raw PDF text:
    - Removes extra whitespace
    - Removes special/broken characters
    - Fixes spacing issues common in PDF extraction
    """
    # Remove non-ASCII broken characters
    text = text.encode("ascii", "ignore").decode()

    # Remove URLs
    text = re.sub(r'http\S+|www\S+', '', text)

    # Remove email addresses
    text = re.sub(r'\S+@\S+', '', text)

    # Remove extra spaces and tabs
    text = re.sub(r'[ \t]+', ' ', text)

    # Remove lines that are just numbers (page numbers)
    text = re.sub(r'^\s*\d+\s*$', '', text, flags=re.MULTILINE)

    # Remove lines shorter than 4 characters (noise/artifacts)
    lines = text.split('\n')
    lines = [line.strip() for line in lines if len(line.strip()) > 4]
    text = '\n'.join(lines)

    # Remove multiple blank lines
    text = re.sub(r'\n{3,}', '\n\n', text)

    return text.strip()
```

 Split Into Sections

```python
def split_into_sections(text):
    """
    Splits cleaned text into sections based on
    common research paper headings.
    Returns a dictionary of section_name: content
    """
    sections = {
        "introduction": "",
        "methodology": "",
        "results": "",
        "conclusion": "",
        "abstract": "",
        "full_text": text   # always keep full text as backup
    }

    # Common heading patterns in research papers
    patterns = {
        "abstract":     r'(abstract)(.*?)(introduction|1\.)',
        "introduction": r'(introduction|1\.?\s+introduction)(.*?)(2\.|method|related)',
        "methodology":  r'(method|methodology|approach|proposed)(.*?)(result|experiment|evaluation)',
        "results":      r'(result|experiment|evaluation)(.*?)(conclusion|discussion|limitation)',
        "conclusion":   r'(conclusion|discussion|limitation)(.*?)($|reference|bibliograph)',
    }

    text_lower = text.lower()

    for section_name, pattern in patterns.items():
        match = re.search(pattern, text_lower, re.DOTALL | re.IGNORECASE)
        if match:
            # Get the actual text at those positions from original text
            start = match.start(2)
            end = match.end(2)
            sections[section_name] = text[start:end].strip()

    return sections
```

```python
all_papers_cleaned = {}
for paper_name, raw_text in all_papers.items():
    all_papers_cleaned[paper_name] = clean_text(raw_text)

test_papers = list(all_papers_cleaned.keys())[:3]

for paper_name in test_papers:
    print(f"\n{'='*60}")
    print(f"📄 {paper_name}")
    print("=" * 60)

    sections = split_into_sections(all_papers_cleaned[paper_name])

    for section, content in sections.items():
        if section == "full_text":
            continue
        if content:
            print(f"\n✅ [{section.upper()}] — {len(content)} chars")
            print(f"   Preview: {content[:150].strip()}...")
        else:
            print(f"\n⚠️  [{section.upper()}] — Not found")
```

Save All Sections

```python
all_papers_sections = {}

for paper_name, cleaned_text in all_papers_cleaned.items():
    sections = split_into_sections(cleaned_text)
    all_papers_sections[paper_name] = sections

print(f"✅ All {len(all_papers_sections)} papers split into sections!")
print("\nSummary:")
for paper_name, sections in all_papers_sections.items():
    found = [s for s in sections if s != "full_text" and sections[s]]
    print(f"  📄 {paper_name[:45]:<45} → sections found: {found}")
```
